## 第 1 步：讀取整個文字檔

```python
# 函式: read() [讀取檔案內容]
# 實體物件: file [file 實體物件]
# 實體方法: file.read() [讀取檔案內容]

# 語法: with as [自動釋放 context manager 資源]
# 物件: file [file 實體物件]
# python 內建的函式: open()
# open(
#     file,             檔案名稱:"sample.txt"[必要]
#     mode='r',         讀取:"r"[有預設值,選項: 'r', 'w', 'x', 'a', 'b', 't', '+']
#     buffering=-1,
#     encoding=None,    處理中文文字:encoding="utf-8"[引數名稱呼叫-可以不用依照引數名稱呼叫]
#     errors=None,
#     newline=None,
#     closefd=True,
#     opener=None
#     )
# 回傳 context manager: with open() as file:

with open("sample.txt", "r", encoding="utf-8") as file:
    content = file.read()   # 實體方法: file.read() 讀取檔案內容
# 離開 with 區塊後, file.close() 會自動被呼叫, 釋放資源

print(content)
```

```python
with open("sample.txt", encoding="utf-8") as file:
    content = file.read()

print(content)
```

## 第 2 步：一行一行讀取

```python
with open("sample.txt", "r", encoding="utf-8") as file:
    for line in file:   # for 迴圈逐行處理
        print(line.rstrip())    # 移除每行結尾的換行符號
```

```python
with open("sample.txt", "r", encoding="utf-8") as file:
    lines = file.readlines()    # 把所有行放進串列

print(lines)
```

## 第 3 步：建立並寫入文字檔

```python
with open("my_notes.txt", "w", encoding="utf-8") as file:   # 在目前資料夾建立 `my_notes.txt`，並寫入兩行文字：
    file.write("這是我的第一個檔案。\n")
    file.write("我正在練習 Python 寫檔。\n")

with open("my_notes.txt", encoding="utf-8") as file:    # 讀取 `my_notes.txt`，確認內容
    content = file.read()
print(content)
```

## 第 4 步：在檔案最後附加內容

```python
with open("my_notes.txt", "a", encoding="utf-8") as file:
    file.write("這一行是後來附加的。\n")

with open("my_notes.txt", encoding="utf-8") as file:
    content = file.read()
print(content)
```

## 第 5 步：使用 pathlib 組合檔案路徑

```python
from pathlib import Path

data_path = Path("sample.txt")
print(data_path.exists())  # 檢查檔案是否存在

content = data_path.read_text(encoding="utf-8")
print(content)
```
